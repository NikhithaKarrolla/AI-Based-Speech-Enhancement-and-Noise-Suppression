```python
import numpy as np
import matplotlib.pyplot as plt

# Sampling rate
sample_rate = 16000

# Duration in seconds
duration = 1

# Time axis
t = np.linspace(
    0,
    duration,
    int(sample_rate * duration),
    endpoint=False
)

# Frequency
frequency = 100

# Generate sine wave
signal = np.sin(2 * np.pi * frequency * t)

print("Number of samples:", len(signal))
print("Sampling rate:", sample_rate)
print("Duration:", duration)
```

```python
plt.figure(figsize=(12, 4))

plt.plot(t[:500], signal[:500])

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("440 Hz Sine Wave")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

frequency = 16000 # Frequency of the sine wave in Hz
duration = 0.02  # 20 milliseconds

sample_rates = [1000, 2000, 4000, 16000]

plt.figure(figsize=(14, 10))

for i, sample_rate in enumerate(sample_rates):

    t = np.arange(0, duration, 1 / sample_rate)

    signal = np.sin(2 * np.pi * frequency * t)

    plt.subplot(4, 1, i + 1)

    plt.plot(t, signal, marker='o')

    plt.title(f"Sampling Rate = {sample_rate} Hz")
    plt.ylabel("Amplitude")
    plt.grid(True)

plt.xlabel("Time (seconds)")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

frequency = 900
sample_rate = 1000
duration = 0.02

t = np.arange(0, duration, 1 / sample_rate)

signal = np.sin(2 * np.pi * frequency * t)

plt.figure(figsize=(12, 4))

plt.stem(t, signal)

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Aliasing: 900 Hz Signal Sampled at 1000 Hz")
plt.grid(True)

plt.show()
```

```python
frequency = 900
sample_rate = 1000
duration = 0.02

# Dense time axis to approximate continuous signal
t_continuous = np.linspace(0, duration, 5000)

continuous_signal = np.sin(
    2 * np.pi * frequency * t_continuous
)

# Actual samples
t_samples = np.arange(
    0,
    duration,
    1 / sample_rate
)

sampled_signal = np.sin(
    2 * np.pi * frequency * t_samples
)

plt.figure(figsize=(14, 5))

plt.plot(
    t_continuous,
    continuous_signal,
    label="Original 900 Hz signal"
)

plt.stem(
    t_samples,
    sampled_signal,
    linefmt='C1-',
    markerfmt='C1o',
    basefmt=' '
)

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Aliasing Demonstration")
plt.grid(True)

plt.show()
```

```python
frequency = 1000
duration = 0.01

sample_rates = [8000, 16000, 22050, 44100, 48000]

plt.figure(figsize=(14, 12))

for i, sample_rate in enumerate(sample_rates):

    t = np.arange(0, duration, 1 / sample_rate)

    signal = np.sin(2 * np.pi * frequency * t)

    plt.subplot(5, 1, i + 1)

    plt.plot(t, signal)

    plt.title(
        f"Frequency = {frequency} Hz, "
        f"Sampling Rate = {sample_rate} Hz"
    )

    plt.grid(True)

plt.tight_layout()
plt.show()
```
